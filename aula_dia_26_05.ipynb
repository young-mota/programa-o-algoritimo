```python
calculadora com as 4 operações
#menu
#opção 1 -somar
#opção 2 -subtrair
#opçao 3 -mutiplicar
#opçaõ 4- divisão
#opção 5 - sair
ef add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y == 0:
        return "Error! Division by zero."
    return x / y

while True:
    print("\n--- Calculadora --- ")
    print("1 - Somar")
    print("2 - Subtrair")
    print("3 - Multiplicar")
    print("4 - Divisão")
    print("5 - Sair")

    choice = input("Escolha uma opção (1/2/3/4/5): ")

    if choice in ('1', '2', '3', '4'):
        try:
            num1 = float(input("Digite o primeiro número: "))
            num2 = float(input("Digite o segundo número: "))
        except ValueError:
            print("Entrada inválida. Por favor, digite números.")
            continue

        if choice == '1':
            print(f"{num1} + {num2} = {add(num1, num2)}")

        elif choice == '2':
            print(f"{num1} - {num2} = {subtract(num1, num2)}")

        elif choice == '3':
            print(f"{num1} * {num2} = {multiply(num1, num2)}")

        elif choice == '4':
            result = divide(num1, num2)
            if isinstance(result, str):
                print(result)
            else:
                print(f"{num1} / {num2} = {result}")

    elif choice == '5':
        print("Saindo da calculadora. Até mais!")
        break
    else:
        print("Opção inválida. Por favor, escolha uma opção de 1 a 5.")


```

```python
#jogo inteligente

#1 par/impar
#2 positivo
#3 maior de 2 numeros
#4 sair
while True:
    print("\n--- Jogo Inteligente ---")
    print("1 - Verificar Par/Ímpar")
    print("2 - Verificar Positivo/Negativo")
    print("3 - Maior de 2 números")
    print("4 - Sair")

    opcao = input("Escolha uma opção (1/2/3/4): ")

    if opcao == '1':
        try:
            num = int(input("Digite um número inteiro: "))
            print(f"O número {num} é {'Par' if num % 2 == 0 else 'Ímpar'}.")
        except ValueError:
            print("Entrada inválida. Digite um número inteiro.")

    elif opcao == '2':
        try:
            num = float(input("Digite um número: "))
            if num > 0:
                print(f"O número {num} é Positivo.")
            elif num < 0:
                print(f"O número {num} é Negativo.")
            else:
                print(f"O número {num} é Zero.")
        except ValueError:
            print("Entrada inválida. Digite um número.")

    elif opcao == '3':
        try:
            num1 = float(input("Digite o primeiro número: "))
            num2 = float(input("Digite o segundo número: "))
            if num1 == num2:
                print("Os números são iguais")
            else:
                print(f"O maior número é: {max(num1, num2)}")
        except ValueError:
            print("Entrada inválida. Digite números.")

    elif opcao == '4':
        print("Saindo do Jogo Inteligente. Até mais!")
        break

    else:
        print("Opção inválida. Escolha uma opção entre 1 e 4.")
```

```python
#senha , o usuario terá 3 tentativas caso erre as 3 imprimir mensagem de acesso negado e tente novamente mais tarde  e encerrar programa


senha_correta = "senha123"
tentativas = 0
max_attempts = 3

while tentativas < max_attempts:
    password_input = input("Digite a senha: ")
    if password_input == senha_correta:
        print("Acesso concedido!")
        break
    else:
        tentativas += 1
        remaining_attempts = max_attempts - tentativas
        if remaining_attempts > 0:
            print(f"Senha incorreta. Você tem {remaining_attempts} tentativa(s) restante(s).")
        else:
            print("Senha incorreta. Nenhuma tentativa restante.")

if tentativas == max_attempts:
    print("Acesso negado. Tente novamente mais tarde.")
```
