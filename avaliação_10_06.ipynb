```python
# Solicita um número inteiro positivo
numero = int(input("Digite um número inteiro positivo: "))

# Inicializa a variável da soma
soma = 0

# Calcula a soma dos números de 1 até o número informado
for i in range(1, numero + 1):
    soma += i

# Exibe o resultado
print(f"A soma dos números de 1 até {numero} é {soma}.")
```

```python
# Solicita um número inteiro positivo
numero = int(input("Digite um número: "))

# Validação
if numero < 0:
    print("Erro: digite um número inteiro positivo.")
else:
    fatorial = 1

    # Calcula o fatorial
    for i in range(1, numero + 1):
        fatorial *= i

    print(f"Fatorial = {fatorial}")
```

```python
# Solicita o valor do saque
valor = int(input("Digite o valor do saque: "))

restante = valor

while restante > 0:
    if restante >= 100:
        notas = restante // 100
        print(f"{notas} nota(s) de R$100")
        restante %= 100

    elif restante >= 50:
        notas = restante // 50
        print(f"{notas} nota(s) de R$50")
        restante %= 50

    elif restante >= 20:
        notas = restante // 20
        print(f"{notas} nota(s) de R$20")
        restante %= 20

    elif restante >= 10:
        notas = restante // 10
        print(f"{notas} nota(s) de R$10")
        restante %= 10

    elif restante >= 5:
        notas = restante // 5
        print(f"{notas} nota(s) de R$5")
        restante %= 5

    elif restante >= 2:
        notas = restante // 2
        print(f"{notas} nota(s) de R$2")
        restante %= 2

    else:
        notas = restante
        print(f"{notas} nota(s) de R$1")
        restante = 0
```
