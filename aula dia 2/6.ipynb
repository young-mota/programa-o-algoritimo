```python
# Fazer um jogo de adivinhação

# O usuário irá digitar um número de 0 a 10 para adivinhar o número secreto
# O usuário terá 3 tentativas para acertar o número
# O programa deve imprimir uma mensagem se o palpite é maior ou menor que o número secreto
# O programa deve imprimir uma mensagem de quente, morno ou frio

# Regras:
# Diferença 1 = Quente
# Diferença 2 = Morno
# Diferença 3 ou mais = Frio

import random

# Gera um número secreto de 0 a 10
numero_secreto = random.randint(0, 10)

tentativas = 0
maximo_tentativas = 3

print("Tente adivinhar o número!")
print("Ele está entre 0 e 10.")

while tentativas < maximo_tentativas:
    try:
        palpite = int(
            input(
                f"Tentativa {tentativas + 1}/{maximo_tentativas}: Digite seu palpite: "
            )
        )

        if not (0 <= palpite <= 10):
            print("Por favor, digite um número entre 0 e 10.")
            continue

        tentativas += 1

        if palpite == numero_secreto:
            print(
                f"Correto! Você adivinhou o número {numero_secreto} em {tentativas} tentativa(s)!"
            )
            break

        if palpite < numero_secreto:
            print("Seu palpite é menor que o número secreto.")
        else:
            print("Seu palpite é maior que o número secreto.")

        diferenca = abs(numero_secreto - palpite)

        if diferenca == 1:
            print("Quente!")
        elif diferenca == 2:
            print("Morno!")
        else:
            print("Frio!")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if tentativas == maximo_tentativas and palpite != numero_secreto:
    print(f"Suas tentativas acabaram. O número secreto era {numero_secreto}.")
```
